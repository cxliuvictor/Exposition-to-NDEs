
# Neural Local Volatility benchmark v5 — Stage 1 serious shared calibration

This notebook is **Stage 1 only**: one serious calibration of the shared neural local-volatility model on **1 March 2024**.

It deliberately does **not** add rolling March calibration, multiple random seeds, warm starts, or the piecewise-$\sigma$ alternative yet. Those come only after this single-date experiment is convincing.

### Core model choices retained from v4

- Same cleaned SPX universe as the classical LV/Heston notebooks.
- 25 OTM calibration quotes per expiry/settlement slice.
- Shared neural local volatility
  $$
  dX_t=X_t\sigma_\theta(t,X_t)\,dW_t,\qquad X_t=S_t/F_0(t).
  $$
- Sabate/Gierjatowicz tamed-Euler simulation.
- Alternating SDE and control-variate optimisation.
- One control-variate network per maturity interval, with one hedge output per selected calibration option.
- Sabate's Xavier-normal **weight-only** initialisation; PyTorch's default biases are left untouched.
- Sabate's Adam learning rate $10^{-3}$, gradient clipping, and scheduled SDE learning-rate reductions at total epochs 500 and 800.
- Raw Monte Carlo pricing error, not the finite-sample control-variate price, is used to judge diffusion-model fit.

### What is new in v5

The goal is to remove the two largest remaining sources of ambiguity from v4: **insufficient optimisation** and **Monte Carlo noise in model selection**.

- 1000 alternating epochs: 500 SDE epochs + 500 control-variate epochs.
- 4096 fresh Monte Carlo paths per SGD step and 10 SGD steps per epoch.
- 16,384 fixed raw paths for sparse convergence monitoring.
- Keep the best 5 checkpoints on that fixed monitor.
- After training, rerank those 5 checkpoints on an **independent 50,000-path sample**.
- Use the independently best checkpoint for the final 100,000-path same-day and next-day evaluations.
- Keep sparse control-variate variance checks and maturity-bucket RMSE monitoring.
- Retain selected-OTM, full-OTM, **and complete-chain** pricing metrics.
- Save training state and CSV diagnostics continuously so a later rendering/kernel failure does not destroy the expensive calibration.

No automatic early stopping or `ReduceLROnPlateau` rule is used. A plateau is **observed**, not automatically acted on, because the full 1000-epoch trajectory and Sabate's scheduled learning-rate changes are themselves informative.

## 2. Load and clean the OptionMetrics data

The cleaning logic is kept aligned with the LV/Heston notebooks:

- SPX (`secid=108105`);
- European, standard 100-multiplier, unadjusted contracts;
- DTE 7–365;
- valid bid/ask and implied volatility;
- $0.05\le |\Delta|\le0.95$;
- relative spread $\le25\%$;
- $|\log(K/F)|\le0.30$;
- elementary European price bounds.

Calibration uses one OTM observation per strike and requires at least 8 observations per expiry/settlement slice.


```python
def find_file(pattern):
    search_dirs = [Path("."), Path("/mnt/data")]
    found = []
    for folder in search_dirs:
        found.extend(sorted(folder.glob(pattern)))
    if not found:
        raise FileNotFoundError(f"Could not find a file matching {pattern!r}")
    return found[0]


FILES = {
    "options": find_file("one_month_spx_option_price.csv"),
    "forwards": find_file("one_month_spx_forward_price.csv"),
    "spot": find_file("one_month_spx_security_price.csv"),
    "rates": find_file("one_month_spx_zero_coupon_yield_curve.csv"),
}

for name, path in FILES.items():
    print(f"{name:8s}: {path}")

options_raw = pd.read_csv(FILES["options"], low_memory=False)
forwards_raw = pd.read_csv(FILES["forwards"], low_memory=False)
spot_raw = pd.read_csv(FILES["spot"], low_memory=False)
rates_raw = pd.read_csv(FILES["rates"], low_memory=False)



SECID = 108105
MIN_DTE, MAX_DTE = 7, 365
MAX_ABS_LOG_MONEYNESS = 0.30
MIN_ABS_DELTA, MAX_ABS_DELTA = 0.05, 0.95
MAX_RELATIVE_SPREAD = 0.25
ATM_TIGHTER_SIDE_BAND = 0.05
MIN_POINTS_PER_SLICE = 8


def prepare_clean_data(options_raw, forwards_raw, spot_raw, rates_raw):
    options = options_raw.copy()
    options["date"] = pd.to_datetime(options["date"])
    options["exdate"] = pd.to_datetime(options["exdate"])
    options["K"] = options["strike_price"] / 1000.0
    options["T_days"] = (options["exdate"] - options["date"]).dt.days
    options["T"] = options["T_days"] / 365.0
    options["bid"] = pd.to_numeric(options["best_bid"], errors="coerce")
    options["ask"] = pd.to_numeric(options["best_offer"], errors="coerce")
    options["mid"] = 0.5 * (options["bid"] + options["ask"])
    options["spread"] = options["ask"] - options["bid"]
    options["relative_spread"] = options["spread"] / options["mid"].replace(0, np.nan)
    options["iv_market"] = pd.to_numeric(options["impl_volatility"], errors="coerce")
    options["delta_market"] = pd.to_numeric(options["delta"], errors="coerce")
    options["abs_delta_market"] = options["delta_market"].abs()

    mask = (
        (options["secid"] == SECID)
        & (options["exercise_style"] == "E")
        & (options["contract_size"] == 100)
        & (options["cfadj"] == 1)
        & options["cp_flag"].isin(["C", "P"])
        & (options["bid"] >= 0)
        & (options["ask"] > options["bid"])
        & (options["mid"] > 0)
        & options["T_days"].between(MIN_DTE, MAX_DTE)
        & options["iv_market"].between(1e-4, 5.0)
        & options["abs_delta_market"].between(MIN_ABS_DELTA, MAX_ABS_DELTA)
        & (options["relative_spread"] <= MAX_RELATIVE_SPREAD)
    )
    options = options.loc[mask].copy()

    spot = spot_raw.loc[spot_raw["secid"] == SECID, ["date", "close"]].copy()
    spot["date"] = pd.to_datetime(spot["date"])
    spot = spot.rename(columns={"close": "S"}).drop_duplicates("date")

    forwards = forwards_raw.loc[forwards_raw["secid"] == SECID].copy()
    forwards["date"] = pd.to_datetime(forwards["date"])
    forwards["expiration"] = pd.to_datetime(forwards["expiration"])
    forwards = forwards.rename(columns={
        "expiration": "exdate",
        "AMSettlement": "am_settlement",
        "ForwardPrice": "F",
    })
    forwards = forwards[
        ["date", "exdate", "am_settlement", "F"]
    ].drop_duplicates(["date", "exdate", "am_settlement"])

    rates = rates_raw.copy()
    rates["date"] = pd.to_datetime(rates["date"])
    rates["days"] = pd.to_numeric(rates["days"], errors="coerce")
    rates["rate_decimal"] = pd.to_numeric(rates["rate"], errors="coerce") / 100.0
    rate_curves = {
        d: (g["days"].to_numpy(float), g["rate_decimal"].to_numpy(float))
        for d, g in rates.dropna().sort_values(["date", "days"]).groupby("date")
    }

    options = options.merge(spot, on="date", how="left").merge(
        forwards,
        on=["date", "exdate", "am_settlement"],
        how="left",
        validate="many_to_one",
    )
    options["r"] = [
        float(np.interp(n, *rate_curves[d])) if d in rate_curves else np.nan
        for d, n in zip(options["date"], options["T_days"])
    ]
    options["D"] = np.exp(-options["r"] * options["T"])
    options["carry"] = np.log(options["F"] / options["S"]) / options["T"]
    options["q"] = options["r"] - options["carry"]
    options["k"] = np.log(options["K"] / options["F"])

    options = options.dropna(subset=["S", "F", "r", "D", "carry", "k"])
    options = options.loc[options["k"].abs() <= MAX_ABS_LOG_MONEYNESS].copy()

    call_lb = options["D"] * np.maximum(options["F"] - options["K"], 0.0)
    call_ub = options["D"] * options["F"]
    put_lb = options["D"] * np.maximum(options["K"] - options["F"], 0.0)
    put_ub = options["D"] * options["K"]
    lb = np.where(options["cp_flag"].eq("C"), call_lb, put_lb)
    ub = np.where(options["cp_flag"].eq("C"), call_ub, put_ub)
    options = options.loc[
        (options["mid"] >= lb - 1e-6) & (options["mid"] <= ub + 1e-6)
    ].copy()

    return options.sort_values(
        ["date", "exdate", "am_settlement", "K", "cp_flag"]
    ).reset_index(drop=True)


def build_otm_surface(quotes):
    idx = [
        "date", "exdate", "am_settlement", "K", "T", "S", "F",
        "r", "q", "carry", "D", "k"
    ]
    vals = [
        "mid", "bid", "ask", "spread", "relative_spread",
        "iv_market", "delta_market", "optionid"
    ]

    wide = quotes.pivot_table(
        index=idx, columns="cp_flag", values=vals, aggfunc="first"
    )
    wide.columns = [f"{a}_{b}" for a, b in wide.columns]
    wide = wide.reset_index()

    strict_call = wide["K"] >= wide["F"]
    near = wide["k"].abs() <= ATM_TIGHTER_SIDE_BAND
    have_c = wide.get("iv_market_C", pd.Series(np.nan, index=wide.index)).notna()
    have_p = wide.get("iv_market_P", pd.Series(np.nan, index=wide.index)).notna()
    c_spread = wide.get("spread_C", pd.Series(np.inf, index=wide.index))
    p_spread = wide.get("spread_P", pd.Series(np.inf, index=wide.index))

    choose_call = strict_call.copy()
    both_near = near & have_c & have_p
    choose_call = np.where(both_near, c_spread <= p_spread, choose_call)
    wide["cp_otm"] = np.where(choose_call, "C", "P")

    for base in vals:
        c = wide.get(f"{base}_C", pd.Series(np.nan, index=wide.index))
        p = wide.get(f"{base}_P", pd.Series(np.nan, index=wide.index))
        wide[f"{base}_otm"] = np.where(choose_call, c, p)

    wide = wide.dropna(subset=["iv_market_otm", "mid_otm", "optionid_otm"]).copy()

    sizes = wide.groupby(["date", "exdate", "am_settlement"]).size()
    valid = sizes[sizes >= MIN_POINTS_PER_SLICE].index

    return (
        wide.set_index(["date", "exdate", "am_settlement"])
        .loc[valid]
        .reset_index()
        .sort_values(["date", "T", "k"])
        .reset_index(drop=True)
    )


quotes_all = prepare_clean_data(options_raw, forwards_raw, spot_raw, rates_raw)
otm = build_otm_surface(quotes_all)

print("Complete cleaned quotes:", len(quotes_all))
print("OTM calibration observations:", len(otm))
print("Trading dates:", otm["date"].nunique())
print("Expiry/settlement slices:", otm.groupby(["date", "exdate", "am_settlement"]).ngroups)
```

## 3. Curves and the 25-per-slice calibration set


```python
@dataclass
class DailyMarketCurve:
    date: pd.Timestamp
    am_settlement: int
    spot: float
    T_nodes: np.ndarray
    logD_nodes: np.ndarray
    logF_nodes: np.ndarray

    def __post_init__(self):
        self._logD = PchipInterpolator(self.T_nodes, self.logD_nodes, extrapolate=True)
        self._logF = PchipInterpolator(self.T_nodes, self.logF_nodes, extrapolate=True)

    def log_discount(self, t):
        return self._logD(np.maximum(np.asarray(t, float), 0.0))

    def log_forward_ratio(self, t):
        return self._logF(np.maximum(np.asarray(t, float), 0.0))

    def discount(self, t0, t1):
        return np.exp(self.log_discount(t1) - self.log_discount(t0))

    def forward(self, t):
        return self.spot * np.exp(self.log_forward_ratio(t))

    def q_integral(self, t0, t1):
        return (
            -(self.log_discount(t1) - self.log_discount(t0))
            -(self.log_forward_ratio(t1) - self.log_forward_ratio(t0))
        )


class CurveBook:
    def __init__(self, spot_raw, forwards_raw, rates_raw):
        self.spot = spot_raw.loc[
            spot_raw["secid"] == SECID, ["date", "close"]
        ].copy()
        self.spot["date"] = pd.to_datetime(self.spot["date"])
        self.spot = self.spot.set_index("date")["close"].to_dict()

        self.forwards = forwards_raw.loc[forwards_raw["secid"] == SECID].copy()
        self.forwards["date"] = pd.to_datetime(self.forwards["date"])
        self.forwards["expiration"] = pd.to_datetime(self.forwards["expiration"])

        self.rates = rates_raw.copy()
        self.rates["date"] = pd.to_datetime(self.rates["date"])
        self.rates["rate_decimal"] = pd.to_numeric(
            self.rates["rate"], errors="coerce"
        ) / 100.0
        self._cache = {}

    def get(self, date, am_settlement):
        date = pd.Timestamp(date)
        am = int(am_settlement)
        key = (date, am)
        if key in self._cache:
            return self._cache[key]

        S0 = float(self.spot[date])

        rg = self.rates.loc[
            self.rates["date"] == date, ["days", "rate_decimal"]
        ].dropna().sort_values("days")
        T_r = rg["days"].to_numpy(float) / 365.0
        logD = -rg["rate_decimal"].to_numpy(float) * T_r

        fg = self.forwards.loc[
            (self.forwards["date"] == date)
            & (self.forwards["AMSettlement"] == am),
            ["expiration", "ForwardPrice"],
        ].copy()
        if len(fg) < 2:
            fg = self.forwards.loc[
                self.forwards["date"] == date,
                ["expiration", "ForwardPrice"],
            ].copy()

        fg["T"] = (fg["expiration"] - date).dt.days / 365.0
        fg = (
            fg.loc[(fg["T"] >= 0) & (fg["ForwardPrice"] > 0)]
            .groupby("T", as_index=False)["ForwardPrice"]
            .mean()
            .sort_values("T")
        )
        T_f = fg["T"].to_numpy(float)
        logF = np.log(fg["ForwardPrice"].to_numpy(float) / S0)
        positive = T_f > 0
        T_f, logF = T_f[positive], logF[positive]

        T_all = np.unique(np.concatenate([[0.0], T_r, T_f]))
        logD_all = np.interp(T_all, np.r_[0.0, T_r], np.r_[0.0, logD])
        logF_all = np.interp(T_all, np.r_[0.0, T_f], np.r_[0.0, logF])

        curve = DailyMarketCurve(date, am, S0, T_all, logD_all, logF_all)
        self._cache[key] = curve
        return curve


curve_book = CurveBook(spot_raw, forwards_raw, rates_raw)



def sample_calibration_day(day, max_per_slice=MAX_QUOTES_PER_SLICE):
    pieces = []
    for _, g in day.groupby(["exdate", "am_settlement"]):
        g = g.sort_values("k")
        if len(g) <= max_per_slice:
            pieces.append(g)
        else:
            pos = np.unique(
                np.round(np.linspace(0, len(g)-1, max_per_slice)).astype(int)
            )
            pieces.append(g.iloc[pos])
    return pd.concat(pieces, ignore_index=True)

def prepare_selected_calibration(date):
    day = otm.loc[otm["date"] == pd.Timestamp(date)].copy()
    selected = sample_calibration_day(day).rename(columns={
        "mid_otm": "target_mid",
        "cp_otm": "target_cp",
        "optionid_otm": "optionid",
    }).reset_index(drop=True)
    selected["contract_id"] = np.arange(len(selected), dtype=int)
    return selected

selected_example = prepare_selected_calibration(EXAMPLE_DATE)
print("Selected contracts:", len(selected_example))
print("Distinct maturity times:", selected_example["T"].nunique())
print("Expiry/settlement slices:",
      selected_example.groupby(["exdate","am_settlement"]).ngroups)
```

## 4. Shared neural LV and Sabate-style interval control variates

The SDE coefficient is shared:
$$
dX_t=X_t\sigma_\theta(t,X_t)\,dW_t,\qquad X_0=1.
$$

The **control variate is not shared through time**. Following Sabate's `Net_timegrid`, there is one hedge network on each maturity interval. Each interval network has one output coordinate for every selected calibration option.

For computational efficiency, an SGD mini-batch only materialises the output coordinates it needs. This is algebraically the same as taking the corresponding coordinates from Sabate's full output vector.

The initialisation below deliberately does **not** zero biases. Sabate's `init_weights` overwrites only `nn.Linear.weight`.


```python
class FeedForwardTrunk(nn.Module):
    def __init__(self, input_dim, hidden_layers, width):
        super().__init__()
        layers, d = [], input_dim
        for _ in range(hidden_layers):
            layers += [nn.Linear(d, width), nn.ReLU()]
            d = width
        self.net = nn.Sequential(*layers)
        self.output_dim = width

    def forward(self, x):
        return self.net(x)

def sabate_weight_initialisation(module):
    # Exact repository behaviour: change weights only, leave biases untouched.
    if isinstance(module, nn.Linear):
        nn.init.xavier_normal_(module.weight.data, gain=1.5)

class SharedNeuralLocalVol(nn.Module):
    def __init__(self):
        super().__init__()
        self.trunk = FeedForwardTrunk(2, SDE_HIDDEN_LAYERS, SDE_WIDTH)
        self.out = nn.Sequential(nn.Linear(SDE_WIDTH, 1), nn.Softplus())
        self.apply(sabate_weight_initialisation)

    def forward(self, t, x):
        if x.ndim == 1:
            x = x[:, None]
        if not torch.is_tensor(t):
            t = torch.full_like(x, float(t))
        elif t.ndim == 0:
            t = torch.full_like(x, float(t.item()))
        elif t.ndim == 1:
            t = t[:, None]
        return self.out(self.trunk(torch.cat([t, x], dim=1)))

class PerOptionHedgeNet(nn.Module):
    def __init__(self, n_options):
        super().__init__()
        self.trunk = FeedForwardTrunk(2, HEDGE_HIDDEN_LAYERS, HEDGE_WIDTH)
        self.output = nn.Linear(HEDGE_WIDTH, int(n_options))
        self.apply(sabate_weight_initialisation)

    def forward_ids(self, t, x, option_ids):
        if x.ndim == 1:
            x = x[:, None]
        if not torch.is_tensor(t):
            t = torch.full_like(x, float(t))
        elif t.ndim == 0:
            t = torch.full_like(x, float(t.item()))
        elif t.ndim == 1:
            t = t[:, None]

        h = self.trunk(torch.cat([t, x], dim=1))
        ids = torch.as_tensor(option_ids, dtype=torch.long, device=h.device)
        W = self.output.weight[ids]
        b = self.output.bias[ids]
        return h @ W.T + b[None, :]

class IntervalHedgeBank(nn.Module):
    def __init__(self, maturity_boundaries, n_options):
        super().__init__()
        self.boundaries = np.asarray(maturity_boundaries, float)
        self.nets = nn.ModuleList([
            PerOptionHedgeNet(n_options) for _ in self.boundaries
        ])

    def interval_index(self, t):
        tt = float(t.reshape(-1)[0].item()) if torch.is_tensor(t) else float(t)
        idx = int(np.searchsorted(self.boundaries, tt + 1e-7, side="right"))
        return min(idx, len(self.nets)-1)

    def forward_ids(self, t, x, option_ids):
        return self.nets[self.interval_index(t)].forward_ids(t, x, option_ids)

def set_requires_grad(module, value):
    for p in module.parameters():
        p.requires_grad_(value)
```

## 5. Time grid, calibration context and Monte Carlo estimators


```python
def build_time_grid(maturities, steps_per_year=BASE_STEPS_PER_YEAR):
    maturities = np.sort(np.unique(np.asarray(maturities, float)))
    max_T = float(maturities[-1])
    base = np.arange(0.0, max_T + 0.5/steps_per_year, 1.0/steps_per_year)
    base = base[base <= max_T + 1e-12]
    grid = np.unique(np.r_[0.0, base, maturities, max_T])
    grid.sort()
    return grid

def nearest_grid_indices(grid, times):
    grid, times = np.asarray(grid), np.asarray(times)
    idx = np.clip(np.searchsorted(grid, times), 0, len(grid)-1)
    left = np.maximum(idx-1, 0)
    idx = np.where(np.abs(grid[left]-times) < np.abs(grid[idx]-times), left, idx)
    if np.max(np.abs(grid[idx]-times)) > 1e-10:
        raise RuntimeError("A listed maturity was not represented exactly on the simulation grid.")
    return idx

@dataclass
class CalibrationContext:
    date: pd.Timestamp
    spot: float
    contracts: pd.DataFrame
    maturity_boundaries: np.ndarray
    grid_np: np.ndarray
    grid: torch.Tensor
    maturity_index: torch.Tensor
    stage_index: np.ndarray
    K: torch.Tensor
    F_T: torch.Tensor
    D_T: torch.Tensor
    target: torch.Tensor
    cp_sign: torch.Tensor
    am: torch.Tensor
    F_grid: torch.Tensor
    D_grid: torch.Tensor

def build_calibration_context(date):
    contracts = prepare_selected_calibration(date)
    date = pd.Timestamp(date)
    boundaries = np.sort(contracts["T"].unique())
    grid_np = build_time_grid(boundaries)
    maturity_idx = nearest_grid_indices(grid_np, contracts["T"].to_numpy(float))
    stage_map = {round(float(T),12): i for i,T in enumerate(boundaries)}
    stage_index = np.array(
        [stage_map[round(float(T),12)] for T in contracts["T"]], dtype=int
    )

    S0 = float(contracts["S"].iloc[0])
    F_grid = np.empty((2, len(grid_np)))
    D_grid = np.empty((2, len(grid_np)))
    for am in [0,1]:
        curve = curve_book.get(date, am)
        F_grid[am] = curve.forward(grid_np)
        D_grid[am] = curve.discount(0.0, grid_np)

    return CalibrationContext(
        date=date,
        spot=S0,
        contracts=contracts,
        maturity_boundaries=boundaries,
        grid_np=grid_np,
        grid=torch.tensor(grid_np, dtype=torch.float32, device=DEVICE),
        maturity_index=torch.tensor(maturity_idx, dtype=torch.long, device=DEVICE),
        stage_index=stage_index,
        K=torch.tensor(contracts["K"].to_numpy(float), dtype=torch.float32, device=DEVICE),
        F_T=torch.tensor(contracts["F"].to_numpy(float), dtype=torch.float32, device=DEVICE),
        D_T=torch.tensor(contracts["D"].to_numpy(float), dtype=torch.float32, device=DEVICE),
        target=torch.tensor(contracts["target_mid"].to_numpy(float), dtype=torch.float32, device=DEVICE),
        cp_sign=torch.tensor(
            np.where(contracts["target_cp"].eq("C"),1.0,-1.0),
            dtype=torch.float32, device=DEVICE
        ),
        am=torch.tensor(
            contracts["am_settlement"].to_numpy(int),
            dtype=torch.long, device=DEVICE
        ),
        F_grid=torch.tensor(F_grid, dtype=torch.float32, device=DEVICE),
        D_grid=torch.tensor(D_grid, dtype=torch.float32, device=DEVICE),
    )

def antithetic_normals(total_paths, n_steps, seed):
    half = (total_paths+1)//2
    gen = torch.Generator(device="cpu")
    gen.manual_seed(int(seed))
    z = torch.randn(half, n_steps, generator=gen, dtype=torch.float32)
    return torch.cat([z,-z], dim=0)[:total_paths].to(DEVICE)

def empirical_payoff_mean(samples, strike_norm, cp_flag):
    x = np.sort(np.asarray(samples,float))
    n = len(x)
    prefix = np.r_[0.0, np.cumsum(x)]
    total = prefix[-1]
    k = np.asarray(strike_norm,float)
    idx = np.searchsorted(x,k,side="right")
    below = prefix[idx]
    above = total-below
    call = (above-k*(n-idx))/n
    put = (k*idx-below)/n
    return np.where(np.asarray(cp_flag)=="C",call,put)

def simulate_estimators(model, hedge, ctx, contract_ids, z, phase):
    ids_np = np.asarray(contract_ids,dtype=int)
    ids = torch.as_tensor(ids_np,dtype=torch.long,device=DEVICE)

    maturity_idx = ctx.maturity_index[ids]
    K, F_T, D_T = ctx.K[ids], ctx.F_T[ids], ctx.D_T[ids]
    cp, am = ctx.cp_sign[ids], ctx.am[ids]

    B, M = z.shape[0], len(ids_np)
    x = torch.ones((B,1),dtype=torch.float32,device=DEVICE)
    cv = torch.zeros((B,M),dtype=torch.float32,device=DEVICE)
    raw = torch.zeros((B,M),dtype=torch.float32,device=DEVICE)
    max_step = int(maturity_idx.max().item())

    for step in range(max_step):
        t = ctx.grid[step]
        dt = ctx.grid[step+1]-ctx.grid[step]
        sqrt_dt = torch.sqrt(dt)

        if phase == "hedge":
            with torch.no_grad():
                sigma = model(t,x)
        else:
            sigma = model(t,x)

        dW = sqrt_dt*z[:,step:step+1]

        active = torch.where(maturity_idx > step)[0]
        if len(active):
            active_ids = ids_np[active.detach().cpu().numpy()]
            if phase == "sde":
                with torch.no_grad():
                    h = hedge.forward_ids(t,x.detach(),active_ids)
            else:
                h = hedge.forward_ids(t,x.detach(),active_ids)

            # Sabate detach: the CV affects the sample estimator but does not
            # create an additional theta-gradient path through the hedge.
            stochastic_piece = x.detach()*sigma.detach()*dW
            discounted_forward = (
                ctx.D_grid[am[active],step]*ctx.F_grid[am[active],step]
            )
            cv[:,active] += h*stochastic_piece*discounted_forward[None,:]

        a = x*sigma
        denominator = 1.0 + torch.abs(x.detach()*sigma.detach())*sqrt_dt
        x = x + a*dW/denominator

        matured = torch.where(maturity_idx == step+1)[0]
        if len(matured):
            S_T = x*F_T[matured][None,:]
            strike = K[matured][None,:]
            call = torch.clamp(S_T-strike,min=0.0)
            put = torch.clamp(strike-S_T,min=0.0)
            raw[:,matured] = (
                torch.where((cp[matured]>0)[None,:],call,put)
                * D_T[matured][None,:]
            )

    return raw, raw-cv

class MaturityBalancedSampler:
    def __init__(self, ctx, per_maturity, seed):
        self.groups = [
            np.where(ctx.stage_index==j)[0]
            for j in range(len(ctx.maturity_boundaries))
        ]
        self.per_maturity = per_maturity
        self.rng = np.random.default_rng(seed)

    def draw(self):
        out = []
        for g in self.groups:
            n = min(self.per_maturity, len(g))
            out.extend(self.rng.choice(g,size=n,replace=False))
        return np.asarray(out,dtype=int)

@torch.no_grad()
def simulate_terminal_x(model, ctx, z):
    x = torch.ones((z.shape[0],1),dtype=torch.float32,device=DEVICE)
    stage_by_grid = {
        int(nearest_grid_indices(ctx.grid_np,[T])[0]): j
        for j,T in enumerate(ctx.maturity_boundaries)
    }
    terminal = {}

    for step in range(len(ctx.grid_np)-1):
        t = ctx.grid[step]
        dt = ctx.grid[step+1]-ctx.grid[step]
        sqrt_dt = torch.sqrt(dt)
        sigma = model(t,x)
        dW = sqrt_dt*z[:,step:step+1]
        a = x*sigma
        x = x + a*dW/(1.0+torch.abs(a)*sqrt_dt)

        if step+1 in stage_by_grid:
            terminal[stage_by_grid[step+1]] = (
                x.squeeze(1).detach().cpu().numpy().copy()
            )
    return terminal

def selected_prices_from_terminal(ctx, terminal):
    rows = []
    for stage in range(len(ctx.maturity_boundaries)):
        ids = np.where(ctx.stage_index==stage)[0]
        g = ctx.contracts.iloc[ids]
        payoff = empirical_payoff_mean(
            terminal[stage],
            g["K"].to_numpy(float)/g["F"].to_numpy(float),
            g["target_cp"].to_numpy(),
        )
        price = g["D"].to_numpy(float)*g["F"].to_numpy(float)*payoff
        tmp = g[["contract_id","optionid","T","k","target_cp","target_mid"]].copy()
        tmp["raw_model_price"] = price
        tmp["raw_error"] = price-tmp["target_mid"].to_numpy(float)
        rows.append(tmp)
    return pd.concat(rows,ignore_index=True)

def raw_selected_diagnostics(model, ctx, z_fixed):
    table = selected_prices_from_terminal(ctx, simulate_terminal_x(model,ctx,z_fixed))
    e = table["raw_error"].to_numpy(float)
    return {
        "rmse": float(np.sqrt(np.mean(e**2))),
        "mae": float(np.mean(np.abs(e))),
        "table": table,
    }

def choose_even_ids(n_total, n):
    ids = np.arange(n_total,dtype=int)
    if n is None or n >= n_total:
        return ids
    pos = np.unique(np.round(np.linspace(0,n_total-1,n)).astype(int))
    return ids[pos]

@torch.no_grad()
def cv_diagnostics(model, hedge, ctx, ids, z):
    raw, cv = simulate_estimators(model,hedge,ctx,ids,z,phase="sde")
    ids_t = torch.as_tensor(ids,dtype=torch.long,device=DEVICE)
    target = ctx.target[ids_t]
    raw_price, cv_price = raw.mean(0), cv.mean(0)
    raw_var, cv_var = raw.var(0,unbiased=False), cv.var(0,unbiased=False)
    return {
        "raw_rmse_subset": float(torch.sqrt(torch.mean((raw_price-target)**2)).cpu()),
        "cv_rmse_subset": float(torch.sqrt(torch.mean((cv_price-target)**2)).cpu()),
        "median_variance_ratio": float(
            torch.median(cv_var/torch.clamp(raw_var,min=1e-12)).cpu()
        ),
    }

calib_ctx = build_calibration_context(EXAMPLE_DATE)
print("Simulation steps:", len(calib_ctx.grid_np)-1)
print("Hedge interval networks:", len(calib_ctx.maturity_boundaries))
```


## 6. Serious alternating training, sparse monitoring, and independent checkpoint reranking

Training follows Sabate's alternating logic:

- **SDE epoch:** minimise squared option-price error using the control-variate estimator.
- **Hedge epoch:** minimise the pathwise estimator variance.

There is **no early stopping**. The SDE learning-rate schedule is the repository schedule: Adam starts at $10^{-3}$, then `MultiStepLR` reduces it at total epochs 500 and 800.

### Monitoring versus optimisation

Every 10 SDE epochs, the diffusion is evaluated on the same 16,384 fixed Brownian paths. This is a monitoring device only; it does not modify the optimiser or learning rate.

The fixed-path monitor is still noisy enough that simply choosing its single minimum can favour a lucky Monte Carlo realization. Therefore v5:

1. keeps the **five** best fixed-path checkpoints;
2. completes all 1000 alternating epochs;
3. reranks those five checkpoints using the same new, independent set of 50,000 antithetic paths;
4. uses the independently best checkpoint for every final pricing/hedging result.

The common 50,000-path reranking sample gives all five candidates the same Monte Carlo shocks, making their comparison much less noisy.

The control-variate variance ratio is checked only every 100 hedge epochs. It is a sanity check on $h$, never a criterion for selecting $\sigma_\theta$.


```python

MATURITY_BUCKETS = [
    ("7-30d", 0.0, 30/365),
    ("31-90d", 30/365, 90/365),
    ("91-180d", 90/365, 180/365),
    ("181-365d", 180/365, 366/365),
]

def maturity_bucket_metrics(table, error_col="raw_error"):
    out = {}
    T = table["T"].to_numpy(float)
    e = table[error_col].to_numpy(float)

    for name, lo, hi in MATURITY_BUCKETS:
        mask = (T > lo) & (T <= hi)
        if np.any(mask):
            out[f"rmse_{name}"] = float(np.sqrt(np.mean(e[mask]**2)))
            out[f"mae_{name}"] = float(np.mean(np.abs(e[mask])))
        else:
            out[f"rmse_{name}"] = np.nan
            out[f"mae_{name}"] = np.nan
    return out


def main_checkpoint_file(date):
    return CACHE_DIR / f"stage1_shared_{pd.Timestamp(date):%Y%m%d}_resume.pt"

def candidate_folder(date):
    folder = CACHE_DIR / f"candidates_{pd.Timestamp(date):%Y%m%d}"
    folder.mkdir(exist_ok=True)
    return folder

def history_csv_file(date):
    return CACHE_DIR / f"training_history_{pd.Timestamp(date):%Y%m%d}_shared_v5.csv"

def candidate_meta_csv_file(date):
    return CACHE_DIR / f"candidate_monitor_scores_{pd.Timestamp(date):%Y%m%d}_shared_v5.csv"

def rerank_csv_file(date):
    return CACHE_DIR / f"candidate_rerank_{pd.Timestamp(date):%Y%m%d}_shared_v5.csv"


def save_resume_checkpoint(
    path, next_epoch, model, hedge, opt_sde, opt_hedge, scheduler,
    history, candidates, training_finished=False, completed=False,
    final_model_state=None, final_hedge_state=None, rerank_records=None
):
    payload = {
        "version": "v5",
        "training_finished": bool(training_finished),
        "completed": bool(completed),
        "next_epoch": int(next_epoch),
        "model_state": model.state_dict(),
        "hedge_state": hedge.state_dict(),
        "opt_sde": opt_sde.state_dict(),
        "opt_hedge": opt_hedge.state_dict(),
        "scheduler": scheduler.state_dict(),
        "history": history,
        "candidates": candidates,
        "final_model_state": final_model_state,
        "final_hedge_state": final_hedge_state,
        "rerank_records": rerank_records,
    }
    torch.save(payload, path)


def maybe_store_top_candidate(model, hedge, candidates, epoch, raw_diag, date):
    """
    Keep only the TOP_K_CHECKPOINTS best models according to the fixed
    16,384-path RAW monitor. The CV price is deliberately irrelevant here.
    """
    score = float(raw_diag["rmse"])
    qualifies = (
        len(candidates) < TOP_K_CHECKPOINTS
        or score < max(c["monitor_rmse"] for c in candidates)
    )
    if not qualifies:
        return candidates

    folder = candidate_folder(date)
    cpath = folder / f"candidate_epoch_{int(epoch):04d}.pt"

    torch.save({
        "epoch": int(epoch),
        "monitor_rmse": score,
        "monitor_mae": float(raw_diag["mae"]),
        "model_state": model.state_dict(),
        "hedge_state": hedge.state_dict(),
    }, cpath)

    candidates = list(candidates) + [{
        "epoch": int(epoch),
        "monitor_rmse": score,
        "monitor_mae": float(raw_diag["mae"]),
        "path": str(cpath),
    }]
    candidates = sorted(candidates, key=lambda x: x["monitor_rmse"])

    while len(candidates) > TOP_K_CHECKPOINTS:
        removed = candidates.pop(-1)
        rpath = Path(removed["path"])
        if rpath.exists():
            try:
                rpath.unlink()
            except OSError:
                pass

    pd.DataFrame(candidates).to_csv(
        candidate_meta_csv_file(date), index=False
    )
    return candidates


@torch.no_grad()
def rerank_candidates_independently(ctx, candidates):
    if not candidates:
        raise RuntimeError("No monitor-selected checkpoints are available to rerank.")

    # Independent of both training paths and the fixed 16,384-path monitor.
    z_rerank = antithetic_normals(
        RERANK_PATHS, len(ctx.grid_np)-1, SEED+5_000
    )

    rows = []
    temp_model = SharedNeuralLocalVol().to(DEVICE)
    set_requires_grad(temp_model, False)
    temp_model.eval()

    for c in tqdm(candidates, desc="Independent checkpoint rerank", unit="model"):
        payload = torch.load(c["path"], map_location=DEVICE)
        temp_model.load_state_dict(payload["model_state"])
        diag = raw_selected_diagnostics(temp_model, ctx, z_rerank)

        rows.append({
            "epoch": int(c["epoch"]),
            "monitor_rmse": float(c["monitor_rmse"]),
            "monitor_mae": float(c["monitor_mae"]),
            "rerank_rmse": float(diag["rmse"]),
            "rerank_mae": float(diag["mae"]),
            "path": c["path"],
        })

    del z_rerank, temp_model
    gc.collect()

    table = pd.DataFrame(rows).sort_values(
        ["rerank_rmse", "monitor_rmse"]
    ).reset_index(drop=True)
    table["independent_rank"] = np.arange(1, len(table)+1)

    table.to_csv(rerank_csv_file(ctx.date), index=False)
    best = table.iloc[0].to_dict()
    return table, best


def train_shared_stage1(ctx):
    path = main_checkpoint_file(ctx.date)

    model = SharedNeuralLocalVol().to(DEVICE)
    hedge = IntervalHedgeBank(ctx.maturity_boundaries, len(ctx.contracts)).to(DEVICE)

    opt_sde = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
    opt_hedge = torch.optim.Adam(hedge.parameters(), lr=LEARNING_RATE)

    # Exact Sabate repository schedule: scheduler is stepped after every
    # alternating epoch, so milestones refer to total epochs.
    scheduler = torch.optim.lr_scheduler.MultiStepLR(
        opt_sde, milestones=[500, 800], gamma=0.2
    )

    start_epoch = 0
    history = []
    candidates = []
    training_finished = False

    if path.exists() and not FORCE_RETRAIN:
        p = torch.load(path, map_location=DEVICE)

        if p.get("completed", False):
            model.load_state_dict(p["final_model_state"])
            hedge.load_state_dict(p["final_hedge_state"])
            model.eval(); hedge.eval()
            set_requires_grad(model, False)
            set_requires_grad(hedge, False)

            print("Loaded completed v5 Stage 1 checkpoint:", path)
            return (
                model,
                hedge,
                pd.DataFrame(p["history"]),
                pd.DataFrame(p["rerank_records"]),
            )

        model.load_state_dict(p["model_state"])
        hedge.load_state_dict(p["hedge_state"])
        opt_sde.load_state_dict(p["opt_sde"])
        opt_hedge.load_state_dict(p["opt_hedge"])
        scheduler.load_state_dict(p["scheduler"])

        start_epoch = int(p["next_epoch"])
        history = list(p["history"])
        candidates = list(p.get("candidates", []))
        training_finished = bool(p.get("training_finished", False))

        if training_finished:
            print("Training already finished; resuming at independent reranking.")
        else:
            print("Resuming v5 training at epoch", start_epoch + 1)

    sampler = MaturityBalancedSampler(
        ctx, QUOTES_PER_MATURITY_PER_STEP, SEED+10
    )

    # The monitoring sample is fixed across the entire run.
    z_monitor = antithetic_normals(
        FIXED_RAW_EVAL_PATHS, len(ctx.grid_np)-1, SEED+100
    )

    # Separate fixed sample for the sparse CV diagnostic.
    z_cv = antithetic_normals(
        CV_EVAL_PATHS, len(ctx.grid_np)-1, SEED+200
    )
    cv_ids = choose_even_ids(len(ctx.contracts), CV_EVAL_CONTRACTS)

    if not training_finished:
        sde_count = sum(1 for e in range(start_epoch) if e % 2 == 0)
        hedge_count = sum(1 for e in range(start_epoch) if e % 2 == 1)

        step_times = []
        session_start = time.time()

        bar = tqdm(
            range(start_epoch, EPOCHS),
            desc="Neural LV v5 Stage 1",
            unit="epoch",
        )

        for epoch in bar:
            phase = "sde" if epoch % 2 == 0 else "hedge"

            set_requires_grad(model, phase == "sde")
            set_requires_grad(hedge, phase == "hedge")

            losses = []

            for _ in range(STEPS_PER_EPOCH):
                ids = sampler.draw()
                z = torch.randn(
                    PATHS_PER_STEP,
                    len(ctx.grid_np)-1,
                    dtype=torch.float32,
                    device=DEVICE,
                )

                opt_sde.zero_grad(set_to_none=True)
                opt_hedge.zero_grad(set_to_none=True)

                tic = time.time()
                raw, cv = simulate_estimators(
                    model, hedge, ctx, ids, z, phase
                )

                ids_t = torch.as_tensor(
                    ids, dtype=torch.long, device=DEVICE
                )

                if phase == "sde":
                    # Same Sabate-style CV-assisted price loss as v4.
                    loss = torch.mean(
                        ((cv.mean(0) - ctx.target[ids_t]) / ctx.spot) ** 2
                    )
                    loss.backward()
                    nn.utils.clip_grad_norm_(
                        model.parameters(), SDE_GRAD_CLIP
                    )
                    opt_sde.step()
                else:
                    loss = torch.var(
                        cv / ctx.spot, dim=0, unbiased=False
                    ).mean()
                    loss.backward()
                    nn.utils.clip_grad_norm_(
                        hedge.parameters(), HEDGE_GRAD_CLIP
                    )
                    opt_hedge.step()

                if len(step_times) < 5:
                    step_times.append(time.time() - tic)

                losses.append(float(loss.detach().cpu()))
                del raw, cv, z

            # Sabate steps the SDE scheduler after every alternating epoch.
            scheduler.step()

            raw_diag = None
            cv_diag = None
            bucket_diag = {}

            if phase == "sde":
                sde_count += 1

                should_monitor = (
                    sde_count == 1
                    or sde_count % EVAL_EVERY_SDE_EPOCHS == 0
                )

                if should_monitor:
                    set_requires_grad(model, False)
                    set_requires_grad(hedge, False)

                    raw_diag = raw_selected_diagnostics(
                        model, ctx, z_monitor
                    )
                    bucket_diag = maturity_bucket_metrics(
                        raw_diag["table"], "raw_error"
                    )

                    candidates = maybe_store_top_candidate(
                        model, hedge, candidates,
                        epoch=epoch+1,
                        raw_diag=raw_diag,
                        date=ctx.date,
                    )

            else:
                hedge_count += 1

                if hedge_count % CV_EVAL_EVERY_HEDGE_EPOCHS == 0:
                    set_requires_grad(model, False)
                    set_requires_grad(hedge, False)
                    cv_diag = cv_diagnostics(
                        model, hedge, ctx, cv_ids, z_cv
                    )

            row = {
                "epoch": epoch+1,
                "phase": phase,
                "train_loss": float(np.mean(losses)),
                "sde_lr": float(opt_sde.param_groups[0]["lr"]),
                "raw_rmse": np.nan if raw_diag is None else raw_diag["rmse"],
                "raw_mae": np.nan if raw_diag is None else raw_diag["mae"],
                "cv_raw_rmse_subset": (
                    np.nan if cv_diag is None else cv_diag["raw_rmse_subset"]
                ),
                "cv_rmse_subset": (
                    np.nan if cv_diag is None else cv_diag["cv_rmse_subset"]
                ),
                "median_variance_ratio": (
                    np.nan if cv_diag is None else cv_diag["median_variance_ratio"]
                ),
                "elapsed_hours_this_session": (time.time()-session_start)/3600,
                **bucket_diag,
            }
            history.append(row)

            post = {
                "phase": phase,
                "loss": f"{row['train_loss']:.2e}",
                "lr": f"{row['sde_lr']:.1e}",
            }
            if raw_diag is not None:
                post["raw_RMSE"] = f"{raw_diag['rmse']:.3f}"
            if cv_diag is not None:
                post["CV/raw_var"] = f"{cv_diag['median_variance_ratio']:.2f}"
            bar.set_postfix(post)

            # One early empirical runtime estimate, after actual work has been measured.
            if len(step_times) == 5 and epoch < 3:
                sec = float(np.mean(step_times))
                remaining_steps = (EPOCHS - epoch - 1) * STEPS_PER_EPOCH
                print(
                    f"Measured {sec:.2f}s per optimisation step; "
                    f"rough remaining training time "
                    f"{sec*remaining_steps/3600:.2f}h "
                    "(before sparse diagnostics and final evaluation)."
                )
                # Do not print this repeatedly.
                step_times.append(np.nan)

            # Crash-resistant lightweight CSV.
            if raw_diag is not None or cv_diag is not None:
                pd.DataFrame(history).to_csv(
                    history_csv_file(ctx.date), index=False
                )

            # Resumable PyTorch state, independent of whether a monitor just ran.
            if (
                (epoch+1) % RESUME_SAVE_EVERY_EPOCHS == 0
                or epoch == EPOCHS-1
            ):
                save_resume_checkpoint(
                    path=path,
                    next_epoch=epoch+1,
                    model=model,
                    hedge=hedge,
                    opt_sde=opt_sde,
                    opt_hedge=opt_hedge,
                    scheduler=scheduler,
                    history=history,
                    candidates=candidates,
                    training_finished=(epoch == EPOCHS-1),
                    completed=False,
                )

            gc.collect()

        training_finished = True

        # Explicitly persist completion of the expensive 1000-epoch stage
        # before starting the independent candidate reranking.
        save_resume_checkpoint(
            path=path,
            next_epoch=EPOCHS,
            model=model,
            hedge=hedge,
            opt_sde=opt_sde,
            opt_hedge=opt_hedge,
            scheduler=scheduler,
            history=history,
            candidates=candidates,
            training_finished=True,
            completed=False,
        )

    del z_monitor, z_cv
    gc.collect()

    # --------------------------------------------------------------
    # Independent reranking of the five monitor-selected candidates.
    # --------------------------------------------------------------
    rerank_table, best = rerank_candidates_independently(
        ctx, candidates
    )

    print("\nIndependent checkpoint reranking:")
    print(
        rerank_table[
            [
                "independent_rank", "epoch",
                "monitor_rmse", "rerank_rmse", "rerank_mae"
            ]
        ].to_string(index=False)
    )

    best_payload = torch.load(best["path"], map_location=DEVICE)
    model.load_state_dict(best_payload["model_state"])
    hedge.load_state_dict(best_payload["hedge_state"])

    model.eval(); hedge.eval()
    set_requires_grad(model, False)
    set_requires_grad(hedge, False)

    final_model_state = copy.deepcopy(model.state_dict())
    final_hedge_state = copy.deepcopy(hedge.state_dict())

    save_resume_checkpoint(
        path=path,
        next_epoch=EPOCHS,
        model=model,
        hedge=hedge,
        opt_sde=opt_sde,
        opt_hedge=opt_hedge,
        scheduler=scheduler,
        history=history,
        candidates=candidates,
        training_finished=True,
        completed=True,
        final_model_state=final_model_state,
        final_hedge_state=final_hedge_state,
        rerank_records=rerank_table.to_dict("records"),
    )

    return model, hedge, pd.DataFrame(history), rerank_table


calibration_date = pd.Timestamp(EXAMPLE_DATE)
dates = sorted(pd.to_datetime(quotes_all["date"].unique()))
next_date = dates[dates.index(calibration_date)+1]

print("Calibration date:", calibration_date.date())
print("Next trading date:", next_date.date())

model, hedge, training_history, rerank_table = train_shared_stage1(
    calib_ctx
)

raw_checks = training_history.dropna(subset=["raw_rmse"]).copy()

columns_to_show = [
    "epoch", "sde_lr", "raw_rmse", "raw_mae",
    "rmse_7-30d", "rmse_31-90d", "rmse_91-180d", "rmse_181-365d"
]
columns_to_show = [c for c in columns_to_show if c in raw_checks.columns]

print("\nRaw-pricing monitor only:")
print(raw_checks[columns_to_show].to_string(index=False))

cv_checks = training_history.dropna(
    subset=["median_variance_ratio"]
)[
    [
        "epoch", "cv_raw_rmse_subset",
        "cv_rmse_subset", "median_variance_ratio"
    ]
]
print("\nSparse control-variate checks:")
print(cv_checks.to_string(index=False))

print("\nSelected final checkpoint after independent reranking:")
print(
    rerank_table.iloc[0][
        ["epoch", "monitor_rmse", "rerank_rmse", "rerank_mae"]
    ].to_string()
)
```


## 7. Independent final same-day evaluation

The model used below is **not** the checkpoint with the smallest 16,384-path monitoring RMSE. It is the best of the top-five monitored candidates after the independent 50,000-path reranking.

The final pricing sample is new again: 100,000 antithetic paths.

Three same-day universes remain visible:

1. selected 25-per-slice calibration contracts;
2. the complete cleaned OTM strip;
3. the complete cleaned call/put chain.

Maturity-bucket errors remain as the main structural sanity check.

A small table of representative contracts also reports paired-antithetic Monte Carlo standard errors, so a small pricing difference can be distinguished from simulation noise.


```python

def tkey(t):
    return round(float(t), 12)


@torch.no_grad()
def simulate_same_day_x_pairs(
    model, maturities, total_paths, spot, bump, chunk_paths, seed
):
    if total_paths % 2 != 0:
        raise ValueError("FINAL_MC_PATHS must be even for paired antithetic evaluation.")

    grid = build_time_grid(maturities)
    maturity_idx = {
        tkey(T): int(nearest_grid_indices(grid, [T])[0])
        for T in np.unique(np.asarray(maturities, float))
    }
    index_to_keys = {}
    for key, idx in maturity_idx.items():
        index_to_keys.setdefault(idx, []).append(key)

    collected = {key: [] for key in maturity_idx}

    gen = torch.Generator(device="cpu")
    gen.manual_seed(seed)

    total_pairs = total_paths // 2
    chunk_pairs = max(1, chunk_paths // 2)
    done_pairs = 0

    bar = tqdm(
        total=total_paths,
        desc="Final same-day MC",
        unit="path",
    )

    while done_pairs < total_pairs:
        n_pairs = min(chunk_pairs, total_pairs-done_pairs)

        z_half = torch.randn(
            n_pairs,
            len(grid)-1,
            generator=gen,
            dtype=torch.float32,
        )
        z = torch.cat([z_half, -z_half], dim=0).to(DEVICE)
        n = 2*n_pairs

        x0 = torch.tensor(
            [1.0, (spot+bump)/spot, (spot-bump)/spot],
            dtype=torch.float32,
            device=DEVICE,
        )
        x = x0[None, :].repeat(n, 1)

        for step in range(len(grid)-1):
            t = float(grid[step])
            dt = float(grid[step+1]-grid[step])
            sqrt_dt = math.sqrt(dt)

            xflat = x.reshape(n*3, 1)
            tflat = torch.full_like(xflat, t)
            sigma = model(tflat, xflat).reshape(n, 3)

            dW = sqrt_dt * z[:, step:step+1]
            a = x * sigma
            x = x + a*dW/(1.0 + torch.abs(a)*sqrt_dt)

            idx = step+1
            if idx in index_to_keys:
                arr = x.detach().cpu().numpy()
                # z ordering is [positive shocks, negative shocks].
                paired = np.stack(
                    [arr[:n_pairs], arr[n_pairs:]],
                    axis=1,
                )  # [pair, antithetic_member, base/up/down]

                for key in index_to_keys[idx]:
                    collected[key].append(paired.copy())

        done_pairs += n_pairs
        bar.update(2*n_pairs)

        del z, x
        gc.collect()

    bar.close()

    return {
        k: np.concatenate(v, axis=0)
        for k, v in collected.items()
    }


def price_rows_from_same_day_pairs(rows, samples_by_T, base_col="mid"):
    out = rows.copy().reset_index(drop=True)

    base_price = np.empty(len(out))
    up_price = np.empty(len(out))
    down_price = np.empty(len(out))

    for T, idx in out.groupby("T").groups.items():
        pos = np.asarray(list(idx), dtype=int)
        g = out.loc[idx]

        pairs = samples_by_T[tkey(T)]
        K_over_F = g["K"].to_numpy(float) / g["F"].to_numpy(float)
        cp = g["cp_flag"].to_numpy()
        scale = g["D"].to_numpy(float) * g["F"].to_numpy(float)

        base = pairs[:, :, 0].reshape(-1)
        up = pairs[:, :, 1].reshape(-1)
        down = pairs[:, :, 2].reshape(-1)

        base_price[pos] = scale * empirical_payoff_mean(base, K_over_F, cp)
        up_price[pos] = scale * empirical_payoff_mean(up, K_over_F, cp)
        down_price[pos] = scale * empirical_payoff_mean(down, K_over_F, cp)

    out["neural_lv_price"] = base_price
    out["neural_lv_delta"] = (
        up_price-down_price
    ) / (2*DELTA_BUMP_SPX_POINTS)

    out["pricing_error"] = (
        out["neural_lv_price"] - out[base_col]
    )
    out["inside_spread"] = out["neural_lv_price"].between(
        out["bid"], out["ask"]
    )
    return out


def paired_option_mc_se(pair_samples, strike_norm, cp_flag, scale):
    """
    Standard error using antithetic PAIRS as the independent observations.
    pair_samples has shape [n_pairs, 2].
    """
    x = np.asarray(pair_samples, float)
    k = float(strike_norm)

    if cp_flag == "C":
        payoff = np.maximum(x-k, 0.0)
    else:
        payoff = np.maximum(k-x, 0.0)

    pair_mean = payoff.mean(axis=1)
    if len(pair_mean) < 2:
        return np.nan

    return float(
        scale
        * np.std(pair_mean, ddof=1)
        / np.sqrt(len(pair_mean))
    )


def representative_mc_uncertainty(rows, samples_by_T, n_contracts=16):
    if len(rows) <= n_contracts:
        chosen = rows.copy()
    else:
        ordered = rows.sort_values(["T", "k"]).reset_index(drop=True)
        pos = np.unique(
            np.round(
                np.linspace(0, len(ordered)-1, n_contracts)
            ).astype(int)
        )
        chosen = ordered.iloc[pos].copy()

    records = []
    for _, r in chosen.iterrows():
        pairs = samples_by_T[tkey(r["T"])][:, :, 0]
        scale = float(r["D"] * r["F"])
        se = paired_option_mc_se(
            pairs,
            float(r["K"]/r["F"]),
            r["cp_flag"],
            scale,
        )
        records.append({
            "optionid": int(r["optionid"]),
            "cp_flag": r["cp_flag"],
            "T_days": int(round(float(r["T"])*365)),
            "K": float(r["K"]),
            "market_mid": float(r["mid"]),
            "model_price": float(r["neural_lv_price"]),
            "absolute_error": float(abs(r["pricing_error"])),
            "MC_standard_error": se,
            "approx_95pct_MC_halfwidth": 1.96*se,
        })

    return pd.DataFrame(records)


day0_full = quotes_all.loc[
    quotes_all["date"] == calibration_date
].copy().reset_index(drop=True)

day0_otm = otm.loc[
    otm["date"] == calibration_date
].copy().reset_index(drop=True)

samples_same_day = simulate_same_day_x_pairs(
    model=model,
    maturities=day0_full["T"].unique(),
    total_paths=FINAL_MC_PATHS,
    spot=float(day0_full["S"].iloc[0]),
    bump=DELTA_BUMP_SPX_POINTS,
    chunk_paths=FINAL_PATH_CHUNK,
    seed=SEED+7_000,
)

same_day_full = price_rows_from_same_day_pairs(
    day0_full, samples_same_day, base_col="mid"
)

otm_eval = day0_otm.rename(columns={
    "mid_otm": "mid",
    "bid_otm": "bid",
    "ask_otm": "ask",
    "cp_otm": "cp_flag",
    "optionid_otm": "optionid",
}).copy()

same_day_otm = price_rows_from_same_day_pairs(
    otm_eval, samples_same_day, base_col="mid"
)

selected_ids = set(
    calib_ctx.contracts["optionid"].astype(int)
)
same_day_selected = same_day_full.loc[
    same_day_full["optionid"].astype(int).isin(selected_ids)
].copy()


def price_summary(df):
    e = df["pricing_error"].to_numpy(float)
    return {
        "n": len(df),
        "RMSE": float(np.sqrt(np.mean(e**2))),
        "MAE": float(np.mean(np.abs(e))),
        "inside_bid_ask": float(df["inside_spread"].mean()),
    }


hierarchy = pd.DataFrame({
    "selected_25_per_slice": price_summary(same_day_selected),
    "full_OTM_strip": price_summary(same_day_otm),
    "complete_clean_chain": price_summary(same_day_full),
}).T

print("Same-day pricing hierarchy:")
print(hierarchy.to_string())


def final_maturity_table(df, error_col="pricing_error"):
    tmp = df.copy()
    tmp["maturity_bucket"] = pd.cut(
        tmp["T"],
        bins=[0, 30/365, 90/365, 180/365, 366/365],
        labels=["7-30d", "31-90d", "91-180d", "181-365d"],
        include_lowest=True,
    )
    return tmp.groupby(
        "maturity_bucket", observed=True
    )[error_col].agg(
        n="size",
        RMSE=lambda x: float(np.sqrt(np.mean(np.asarray(x, float)**2))),
        MAE=lambda x: float(np.mean(np.abs(np.asarray(x, float)))),
    )


selected_by_T = final_maturity_table(same_day_selected)
full_otm_by_T = final_maturity_table(same_day_otm)
full_chain_by_T = final_maturity_table(same_day_full)

print("\nSelected-strip same-day error by maturity:")
print(selected_by_T.to_string())

print("\nFull-OTM same-day error by maturity:")
print(full_otm_by_T.to_string())

print("\nComplete-chain same-day error by maturity:")
print(full_chain_by_T.to_string())


mc_uncertainty_same_day = representative_mc_uncertainty(
    same_day_full, samples_same_day, n_contracts=16
)

print("\nRepresentative paired-antithetic Monte Carlo uncertainty:")
print(mc_uncertainty_same_day.to_string(index=False))

print(
    "\nRepresentative median MC standard error: $",
    round(float(mc_uncertainty_same_day["MC_standard_error"].median()), 4)
)

# Save the expensive same-day results before moving to next-day evaluation.
stem = f"{calibration_date:%Y%m%d}_shared_stage1_v5"

hierarchy.to_csv(
    CACHE_DIR / f"same_day_hierarchy_{stem}.csv"
)
selected_by_T.to_csv(
    CACHE_DIR / f"selected_maturity_error_{stem}.csv"
)
full_otm_by_T.to_csv(
    CACHE_DIR / f"full_otm_maturity_error_{stem}.csv"
)
full_chain_by_T.to_csv(
    CACHE_DIR / f"full_chain_maturity_error_{stem}.csv"
)
mc_uncertainty_same_day.to_csv(
    CACHE_DIR / f"same_day_mc_uncertainty_{stem}.csv",
    index=False,
)
same_day_full.to_csv(
    CACHE_DIR / f"same_day_complete_chain_prices_{stem}.csv",
    index=False,
)

del samples_same_day
gc.collect()
```


## 8. Frozen next-day pricing and the common one-day delta hedge

The independently reranked day-$t$ model is now frozen.

For 4 March:

- use the new spot and new forward/discount curves;
- reduce remaining maturity by the calendar gap;
- advance the model clock;
- evaluate the frozen day-1 neural volatility in its original calibration coordinate.

The headline hedge remains the model delta on the complete identical-contract universe.


```python

def build_evaluation_pairs(quotes):
    trading_dates = sorted(quotes["date"].unique())
    next_map = {
        trading_dates[i]: trading_dates[i+1]
        for i in range(len(trading_dates)-1)
    }

    left = quotes.copy()
    left["next_date"] = left["date"].map(next_map)

    cols = [
        "date", "optionid", "cp_flag",
        "K", "exdate", "am_settlement",
        "mid", "bid", "ask",
        "S", "F", "D", "r", "q",
        "carry", "T", "k",
    ]

    right = quotes[cols].rename(columns={
        "date": "next_date",
        "mid": "mid_next",
        "bid": "bid_next",
        "ask": "ask_next",
        "S": "S_next",
        "F": "F_next",
        "D": "D_next",
        "r": "r_next",
        "q": "q_next",
        "carry": "carry_next",
        "T": "T_next",
        "k": "k_next",
    })

    return left.merge(
        right,
        on=[
            "next_date", "optionid",
            "cp_flag", "K", "exdate",
            "am_settlement",
        ],
        how="inner",
        validate="many_to_one",
    )


evaluation_pairs = build_evaluation_pairs(quotes_all)

transition_pairs = evaluation_pairs.loc[
    (evaluation_pairs["date"] == calibration_date)
    & (evaluation_pairs["next_date"] == next_date)
].copy().reset_index(drop=True)

print("Identical-contract pairs:", len(transition_pairs))


@torch.no_grad()
def simulate_next_day_y_pairs(
    model, curve0, curve1, elapsed, maturities_remaining,
    total_paths, chunk_paths, seed
):
    if total_paths % 2 != 0:
        raise ValueError("FINAL_MC_PATHS must be even for paired antithetic evaluation.")

    grid = build_time_grid(maturities_remaining)

    maturity_idx = {
        tkey(T): int(nearest_grid_indices(grid, [T])[0])
        for T in np.unique(np.asarray(maturities_remaining, float))
    }

    index_to_keys = {}
    for key, idx in maturity_idx.items():
        index_to_keys.setdefault(idx, []).append(key)

    collected = {key: [] for key in maturity_idx}

    gen = torch.Generator(device="cpu")
    gen.manual_seed(seed)

    total_pairs = total_paths // 2
    chunk_pairs = max(1, chunk_paths // 2)
    done_pairs = 0

    bar = tqdm(
        total=total_paths,
        desc=f"Next-day MC AM={curve1.am_settlement}",
        unit="path",
    )

    while done_pairs < total_pairs:
        n_pairs = min(chunk_pairs, total_pairs-done_pairs)

        z_half = torch.randn(
            n_pairs,
            len(grid)-1,
            generator=gen,
            dtype=torch.float32,
        )
        z = torch.cat([z_half, -z_half], dim=0).to(DEVICE)
        n = 2*n_pairs

        y = torch.ones(
            (n,1),
            dtype=torch.float32,
            device=DEVICE,
        )

        for step in range(len(grid)-1):
            v = float(grid[step])
            dt = float(grid[step+1]-grid[step])
            sqrt_dt = math.sqrt(dt)

            abs_t = elapsed + v
            Fnew = float(curve1.forward(v))
            Fold = float(curve0.forward(abs_t))

            x_old = y * (Fnew/Fold)
            sigma = model(
                torch.full_like(y, abs_t),
                x_old,
            )

            dW = sqrt_dt * z[:, step:step+1]
            a = y * sigma
            y = y + a*dW/(1.0 + torch.abs(a)*sqrt_dt)

            idx = step+1
            if idx in index_to_keys:
                arr = y.squeeze(1).detach().cpu().numpy()
                paired = np.stack(
                    [arr[:n_pairs], arr[n_pairs:]],
                    axis=1,
                )  # [pair, antithetic_member]

                for key in index_to_keys[idx]:
                    collected[key].append(paired.copy())

        done_pairs += n_pairs
        bar.update(2*n_pairs)

        del z, y
        gc.collect()

    bar.close()

    return {
        k: np.concatenate(v, axis=0)
        for k, v in collected.items()
    }


def price_next_day_from_pairs(pairs, samples_by_am):
    out = pairs.copy().reset_index(drop=True)
    price = np.empty(len(out))

    for (am, T), idx in out.groupby(
        ["am_settlement", "T_next"]
    ).groups.items():
        pos = np.asarray(list(idx), dtype=int)
        g = out.loc[idx]

        pair_samples = samples_by_am[int(am)][tkey(T)]
        samples = pair_samples.reshape(-1)

        strike_norm = (
            g["K"].to_numpy(float)
            / g["F_next"].to_numpy(float)
        )

        payoff = empirical_payoff_mean(
            samples,
            strike_norm,
            g["cp_flag"].to_numpy(),
        )

        price[pos] = (
            g["D_next"].to_numpy(float)
            * g["F_next"].to_numpy(float)
            * payoff
        )

    out["model_price_next"] = price
    out["next_day_pricing_error"] = (
        price - out["mid_next"].to_numpy(float)
    )
    out["next_day_inside_bid_ask"] = (
        out["model_price_next"].between(
            out["bid_next"], out["ask_next"]
        )
    )
    return out


elapsed = (next_date-calibration_date).days / 365.0
samples_next_by_am = {}

for am in sorted(
    transition_pairs["am_settlement"].unique()
):
    g = transition_pairs.loc[
        transition_pairs["am_settlement"] == am
    ]

    curve0 = curve_book.get(
        calibration_date, int(am)
    )
    curve1 = curve_book.get(
        next_date, int(am)
    )

    samples_next_by_am[int(am)] = simulate_next_day_y_pairs(
        model=model,
        curve0=curve0,
        curve1=curve1,
        elapsed=elapsed,
        maturities_remaining=g["T_next"].unique(),
        total_paths=FINAL_MC_PATHS,
        chunk_paths=FINAL_PATH_CHUNK,
        seed=SEED+8_000+int(am),
    )


next_day_results = price_next_day_from_pairs(
    transition_pairs, samples_next_by_am
)

e = next_day_results[
    "next_day_pricing_error"
].to_numpy(float)

print("Next-day n:", len(next_day_results))
print(
    "Next-day RMSE: $",
    round(float(np.sqrt(np.mean(e**2))), 4),
)
print(
    "Next-day MAE:  $",
    round(float(np.mean(np.abs(e))), 4),
)
print(
    "Next-day inside bid-ask:",
    round(
        float(
            next_day_results[
                "next_day_inside_bid_ask"
            ].mean()
        ),
        4,
    ),
)


def representative_next_day_mc_uncertainty(
    rows, samples_by_am, n_contracts=16
):
    if len(rows) <= n_contracts:
        chosen = rows.copy()
    else:
        ordered = rows.sort_values(
            ["T_next", "k_next"]
        ).reset_index(drop=True)
        pos = np.unique(
            np.round(
                np.linspace(
                    0, len(ordered)-1, n_contracts
                )
            ).astype(int)
        )
        chosen = ordered.iloc[pos].copy()

    records = []

    for _, r in chosen.iterrows():
        pair_samples = samples_by_am[
            int(r["am_settlement"])
        ][tkey(r["T_next"])]

        scale = float(
            r["D_next"] * r["F_next"]
        )

        se = paired_option_mc_se(
            pair_samples,
            float(r["K"]/r["F_next"]),
            r["cp_flag"],
            scale,
        )

        records.append({
            "optionid": int(r["optionid"]),
            "cp_flag": r["cp_flag"],
            "T_next_days": int(
                round(float(r["T_next"])*365)
            ),
            "K": float(r["K"]),
            "market_mid_next": float(r["mid_next"]),
            "model_price_next": float(r["model_price_next"]),
            "absolute_error": float(
                abs(r["next_day_pricing_error"])
            ),
            "MC_standard_error": se,
            "approx_95pct_MC_halfwidth": 1.96*se,
        })

    return pd.DataFrame(records)


mc_uncertainty_next = (
    representative_next_day_mc_uncertainty(
        next_day_results,
        samples_next_by_am,
        n_contracts=16,
    )
)

print(
    "\nRepresentative next-day median MC standard error: $",
    round(
        float(
            mc_uncertainty_next[
                "MC_standard_error"
            ].median()
        ),
        4,
    ),
)


# ------------------------------------------------------------------
# Common one-day model-delta hedge
# ------------------------------------------------------------------
delta_lookup = same_day_full[
    ["optionid", "neural_lv_delta"]
].drop_duplicates("optionid")

hedge_results = next_day_results.merge(
    delta_lookup,
    on="optionid",
    how="left",
    validate="many_to_one",
)


def one_day_hedging_error(rows, hedge_ratio):
    rows = rows.copy().reset_index(drop=True)
    hedge_ratio = np.asarray(hedge_ratio, float)

    q_integral = np.empty(len(rows))
    cash_interest = np.empty(len(rows))

    dt = (
        rows["next_date"].iloc[0]
        - rows["date"].iloc[0]
    ).days / 365.0

    for am, idx in rows.groupby(
        "am_settlement"
    ).groups.items():
        pos = np.asarray(list(idx), dtype=int)

        curve0 = curve_book.get(
            rows.loc[idx, "date"].iloc[0],
            int(am),
        )

        q_integral[pos] = float(
            curve0.q_integral(0.0, dt)
        )

        cash0 = (
            rows.loc[idx, "mid"].to_numpy(float)
            - hedge_ratio[pos]
            * rows.loc[idx, "S"].to_numpy(float)
        )

        cash_interest[pos] = (
            cash0
            * (
                1.0
                / float(curve0.discount(0.0, dt))
                - 1.0
            )
        )

    dividend_income = (
        hedge_ratio
        * rows["S"].to_numpy(float)
        * q_integral
    )

    return (
        rows["mid_next"].to_numpy(float)
        - rows["mid"].to_numpy(float)
        - hedge_ratio
        * (
            rows["S_next"].to_numpy(float)
            - rows["S"].to_numpy(float)
        )
        - dividend_income
        - cash_interest
    )


hedge_results["delta_hedging_error"] = (
    one_day_hedging_error(
        hedge_results,
        hedge_results["neural_lv_delta"],
    )
)

hedge_results["unhedged_change"] = (
    hedge_results["mid_next"]
    - hedge_results["mid"]
)


def hedge_summary(error, unhedged):
    error = np.asarray(error, float)
    unhedged = np.asarray(unhedged, float)

    mae = float(np.mean(np.abs(error)))
    umae = float(np.mean(np.abs(unhedged)))

    return {
        "n": len(error),
        "RMSE": float(np.sqrt(np.mean(error**2))),
        "MAE": mae,
        "bias": float(np.mean(error)),
        "unhedged_MAE": umae,
        "improvement": 1-mae/umae,
    }


print("\nModel-delta one-day hedge:")
print(
    pd.Series(
        hedge_summary(
            hedge_results["delta_hedging_error"],
            hedge_results["unhedged_change"],
        )
    ).to_string()
)


# Persist next-day results before the secondary learned-CV hedge section.
next_day_results.to_csv(
    CACHE_DIR / f"next_day_prices_{stem}.csv",
    index=False,
)
hedge_results.to_csv(
    CACHE_DIR / f"delta_hedge_results_{stem}.csv",
    index=False,
)
mc_uncertainty_next.to_csv(
    CACHE_DIR / f"next_day_mc_uncertainty_{stem}.csv",
    index=False,
)

del samples_next_by_am
gc.collect()
```

## 9. Secondary Sabate-native learned hedge

The headline hedge remains the model delta on the complete identical-contract universe.

For the selected calibration contracts only, the learned control-variate hedge is also evaluated over the same observed one-day move. This is retained as a secondary diagnostic because the learned hedge is a central component of Sabate's variance-reduction method.


```python
@torch.no_grad()
def learned_cv_hedges_selected(hedge,ctx):
    values=np.empty(len(ctx.contracts))
    x0=torch.ones((1,1),dtype=torch.float32,device=DEVICE)
    t0=torch.zeros_like(x0)
    for start in range(0,len(ctx.contracts),256):
        ids=np.arange(start,min(start+256,len(ctx.contracts)))
        values[ids]=hedge.forward_ids(t0,x0,ids).squeeze(0).cpu().numpy()
    return values

selected_hedges=calib_ctx.contracts[["optionid","contract_id"]].copy()
selected_hedges["learned_cv_hedge"]=learned_cv_hedges_selected(
    hedge,calib_ctx
)

selected_pairs=hedge_results.merge(
    selected_hedges[["optionid","learned_cv_hedge"]],
    on="optionid",how="inner",validate="many_to_one"
).reset_index(drop=True)

selected_pairs["cv_hedging_error"]=one_day_hedging_error(
    selected_pairs,selected_pairs["learned_cv_hedge"]
)
selected_pairs["delta_hedging_error_selected"]=one_day_hedging_error(
    selected_pairs,selected_pairs["neural_lv_delta"]
)

comparison=pd.DataFrame({
    "model_delta_selected":hedge_summary(
        selected_pairs["delta_hedging_error_selected"],
        selected_pairs["unhedged_change"]
    ),
    "learned_CV_selected":hedge_summary(
        selected_pairs["cv_hedging_error"],
        selected_pairs["unhedged_change"]
    ),
}).T

print("Selected-contract hedge comparison:")
print(comparison.to_string())
```

### How to interpret Stage 1

The first calibration question is the **full OTM same-day RMSE**, because that is the market strip used to identify the neural-LV model. The complete-chain RMSE remains a headline result and should always be reported alongside it.

A useful **Case A** outcome is approximately:

$$
\text{full-OTM same-day RMSE}\lesssim \$4\text{–}5,
$$

with:

- selected and full-OTM RMSE close to one another;
- no maturity bucket failing catastrophically;
- the 50k independent checkpoint ranking broadly agreeing with the final 100k evaluation;
- control-variate variance ratio comfortably below 1;
- sensible one-day delta hedging;
- no obvious collapse in frozen next-day pricing.

If the full-OTM fit instead cleanly plateaus around \$5\text{–}7 despite the complete 1000-epoch trajectory and the scheduled learning-rate reductions, that is evidence for the **next architectural experiment**: Sabate's original piecewise-$\sigma$ construction.

Multiple seeds, warm starts, and the full March roll are intentionally **not part of v5**. They only become relevant after this Stage 1 calibration is accepted.
