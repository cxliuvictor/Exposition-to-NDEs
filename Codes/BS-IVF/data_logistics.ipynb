```python
import pandas as pd

df = pd.read_csv(
    "spx_option_prices_0.csv",
    parse_dates = ["date", "exdate"]
)
# print(df.shape)
# print(df.columns.tolist())
# print(df.tail(10))

n_total = len(df)
n_missing = df["impl_volatility"].isna().sum()

print(f"Total rows: {n_total:,}")
print(f"Missing implied volatility: {n_missing:,}")
print(f"Missing percentage: {100 * n_missing / n_total:.2f}%")

columns_to_view_nan = [
    "date",
    "exdate",
    "strike_price",
    "best_bid",
    "best_offer",
    "impl_volatility"
]

available_columns = [c for c in columns_to_view_nan if c in df.columns]

print(
    df.loc[df["impl_volatility"].isna(), available_columns]
      .head(10)
      .to_string(index=False)
)
```

```python
ivf_df = df.dropna(subset=["impl_volatility"]).copy()
ivf_df = ivf_df.rename(
    columns={
        "strike_price": "strike",
        "best_bid": "bid",
        "best_offer": "ask",
        "impl_volatility": "market_iv",
        "forward_price": "f_price"
    }
)

col_to_view = [
    "date",
    "exdate",
    "strike",
    "bid",
    "ask",
    "market_iv",
    "f_price"
]
print(ivf_df[col_to_view].head(20))
```
